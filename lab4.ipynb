```python
import random

# --------------------------------------------------
# Part A - Machine Initialization
# --------------------------------------------------

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# --------------------------------------------------
# Part B - Initialize N, R and Q
# --------------------------------------------------

# Number of times each machine is selected
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward obtained from each machine
R = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Estimated Q-value of each machine
Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

# --------------------------------------------------
# Given Reward Patterns
# --------------------------------------------------

reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward to use next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# Total reward received
total_reward = 0


# --------------------------------------------------
# Part C - Epsilon-Greedy Strategy
# --------------------------------------------------

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Exploration
    if r < epsilon:

        selected_machine = random.choice(machines)
        decision = "Exploration"

    # Exploitation
    else:

        max_q = max(Q.values())

        # Find all machines having maximum Q-value
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        # Randomly select among best machines
        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # --------------------------------------------------
    # Part D - Reward Collection
    # --------------------------------------------------

    # Check whether rewards are available
    if reward_index[selected_machine] >= len(
        reward_pattern[selected_machine]
    ):
        print("\nNo more rewards available for Machine",
              selected_machine)
        break

    # Get next reward from the given pattern
    reward = reward_pattern[selected_machine][
        reward_index[selected_machine]
    ]

    # Move to next reward
    reward_index[selected_machine] += 1

    print("\n----------------------------------")
    print("Play:", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)
    print("Reward obtained:", reward)

    # --------------------------------------------------
    # Part E - Q-Value Update
    # --------------------------------------------------

    # Increase participation count
    N[selected_machine] += 1

    # Add reward to total reward of machine
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine])
        / N[selected_machine]
    )

    # Add to overall total reward
    total_reward += reward

    print("Updated N:", N)
    print("Updated R:", R)

    print(
        "Updated Q:",
        {m: round(Q[m], 2) for m in machines}
    )


# --------------------------------------------------
# Part F - Final Results
# --------------------------------------------------

print("\n\n========================================")
print("FINAL RESULTS")
print("========================================")

print(
    "\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)"
)

for machine in machines:

    print(
        machine,
        "\t",
        N[machine],
        "\t\t",
        R[machine],
        "\t\t",
        round(Q[machine], 2)
    )

# --------------------------------------------------
# Total Reward
# --------------------------------------------------

print("\nTotal Reward after 10 plays:", total_reward)

# --------------------------------------------------
# Best Machine
# --------------------------------------------------

best_machine = max(Q, key=Q.get)

print("Best Machine:", best_machine)
print("Best Q-Value:", round(Q[best_machine], 2))
```
